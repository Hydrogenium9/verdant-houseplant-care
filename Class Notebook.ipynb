# Entry 1 — Verdant: Houseplant Care & Swapping

**Date:** October 5, 2026

## Goal
Help new houseplant owners follow clear, consistent care guidance and find local plant swaps.

## Work completed
Built a responsive React/Vite dashboard with sample plant profiles, practical watering guides, a care checklist, add/search interactions, and a neighborhood swap board.

## Design and technical decisions
- Keep one consolidated care guide per plant to reduce conflicting advice.
- Set Vite's base path from the GitHub repository name during Actions builds.
- Add a GitHub Actions workflow to build and publish `dist` to GitHub Pages on pushes to `main`.

## Validation
`npm run lint` and `npm run build` pass. Browser checks covered plant care profiles, watering, adding a plant, and the mobile layout.

## Next steps
Push the project to a new GitHub repository, set Pages' source to GitHub Actions, wait for the first workflow run, and verify the published URL.
